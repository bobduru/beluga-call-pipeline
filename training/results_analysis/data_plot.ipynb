```python
import pandas as pd

labels_df = pd.read_csv("../../data/Verified_dataset/labels/labels_merged.csv")
```

```python
labels_df.columns
```

```python
labels_df["Boat"] = labels_df["Boat"].fillna(0)
```

```python
def get_calls_table(labels_df, group_col="Site"):

    call_types = ["HFPC", "BBPC", "ECHO", "Whistle"]

    call_counts = labels_df.groupby(group_col)["Call"].value_counts().unstack(fill_value=0)
    call_counts = call_counts.rename(columns={1: "Present", 0: "Absent"})[["Present", "Absent"]]

    call_type_counts = labels_df.groupby(group_col)[call_types].sum().astype(int)

    table = pd.concat([call_counts, call_type_counts], axis=1)
    table.loc["Total"] = table.sum()
    table = table.astype(int)
    return table
```

```python
get_calls_table(labels_df)  
```

```python
get_calls_table(labels_df, "Boat")  
```
